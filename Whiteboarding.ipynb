```python
def findHighest(*args) -> int:
   
    if len(args) == 1 and isinstance(args[0], list):
        numbers = args[0]
    else:
        numbers = args


    highest = numbers[0]
    
    
    for num in numbers:
        if num > highest:
            highest = num
            
    return highest


print(findHighest(5, 8, 1))    
print(findHighest([4, 1, -3]))  
```

```python
# a = int1 10, 10 b = int2 20, 10

def sum2Integers(a: int, b: int) -> int:

#if equal multuply by 3. 

    if a == b:
        return 3 * (a + b)
    return a + b

print(sum2Integers(10, 20))
print(sum2Integers(10, 10))
```

```python
def countVowels(word):
    count = 0
    for letter in word.lower():
        if letter in 'aeiou':
            count += 1
    return count

print(countVowels('hello'))  # 2
print(countVowels('why'))    # 0
print(countVowels('aeiou'))  # 5 
```

```python
def findLargest(numbers):          # level 0: the function starts
    biggest = numbers[0]           # level 1: inside the function

    for num in numbers:            # level 1: inside the function
        if num > biggest:          # level 2: inside the loop
            biggest = num          # level 3: inside the if

    return biggest                 # level 1: inside the function, after the loop

print(findLargest([3, 5, 7, 9, 10]))
```
