## Rock, Paper, Scissors

```python
import random


choices = ['rock', 'paper', 'scissors']

player = input('Choose rock, paper, or scissors: ').lower()
computer = random.choice(choices)

while player not in choices and player != 'q':
    print('invalid input')          
    player = input('Choose rock, paper, scissors, or q to quit: ').lower()

# if the player chooses a non choice try again or tie

if player == 'q':
    print("Game Cancelled")
elif player == computer:
    print("its a tie!")

# The player chose rock, () in elif not needed but still works with :

elif (player =='rock' and computer == 'scissors'):
    print("computer wins!")
elif (player =='rock' and computer == 'paper'):
    print("player wins!")

# The player chose paper

elif (player =='paper' and computer == 'scissors'):
    print("computer wins!")
elif (player =='paper' and computer == 'rock'):
    print("player wins!") 

# player chose scissors 
elif (player =='scissors' and computer == 'rock'):
    print("computer wins!")
elif player =='scissors' and computer == 'paper':
    print("player wins!") 

print("You chose:", player)
print("Not so Master chose", computer)      
```
